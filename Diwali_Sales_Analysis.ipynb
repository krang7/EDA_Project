```python
# import python libraries

import numpy as np # this is used for a number of mathematical operations
import pandas as pd #this is used for data processing and CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt # visualizing data matplotlib inline
import seaborn as sns #this is used for plotting graphs
```

```python
# import csv file
df = pd.read_csv('Diwali_Sales_Data.csv', encoding= 'unicode_escape') #this will read the csv file
```

```python
df.head() #this will print the first 5 rows of the Data Frame
```

```python
df.info() #this will print the summary of the Data Frame
```

```python
#drop unrelated/blank columns
df.drop(['Status', 'unnamed1'], axis=1, inplace=True) #this will drop the columns 'Status' and 'unnamed1'
```

```python
#check for null values
pd.isnull(df).sum() #this will print the number of null values in each column
```

```python
# drop null values
df.dropna(inplace=True) #this will drop the null values
```

```python
# change data type
df['Amount'] = df['Amount'].astype('int') #this will change the data type of the column 'Amount' to integer
```

```python
df['Amount'].dtypes #this will print the data type of the column 'Amount'
```

```python
df.columns #this will print the columns of the Data Frame
```

```python
#rename column
df.rename(columns= {'Marital_Status':'Shaadi'}) #this will rename the column 'Marital_Status' to 'Shaadi'
```

```python
# describe() method returns description of the data in the DataFrame (i.e. count, mean, std, etc)
df.describe() #this will print the summary statistics of the Data Frame
```

```python
# use describe() for specific columns
df[['Age', 'Orders', 'Amount']].describe() #this will print the summary statistics of the columns 'Age', 'Orders' and 'Amount'
```

# Exploratory Data Analysis

### Gender

# plotting a bar chart for Gender and it's count



```python
ax = sns.countplot(x = 'Gender',data = df) #this will plot the count

for bars in ax.containers: #this will add the count labels
    ax.bar_label(bars) #this will add the count labels
```

```python
# plotting a bar chart for gender vs total amount


sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False) #this will group the data by gender and sum the amount

sns.barplot(x = 'Gender',y= 'Amount' ,data = sales_gen) #this will plot the bar chart 
```

*From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men*

```python
### Age

ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')# this function is used to plot a bar chart

for bars in ax.containers:#This loop is used to display the count of each bar in the bar chart
        ax.bar_label(bars)# this function is used to display the count of each bar in the bar chart
```

```python
# Total Amount vs Age Group
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False) #this will group the data by age group and sum the amount

sns.barplot(x = 'Age Group',y= 'Amount' ,data = sales_age) #this will plot the bar chart
```

*From above graphs we can see that most of the buyers are of age group between 26-35 yrs female*

```python
### State

# total number of orders from top 10 states

sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10) #this will group the data by state and sum the orders

sns.set(rc={'figure.figsize':(15,5)})   #this will set the figure size of the plot for the top 10 states s 
sns.barplot(data = sales_state, x = 'State',y= 'Orders') #this will plot the bar chart with the top 10 states of the order list 
```

```python
# total amount/sales from top 10 states

sales_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10) #this will group the data by state and sum the amount

sns.set(rc={'figure.figsize':(15,5)}) #this will set the figure size of the plot for the top 10 states
sns.barplot(data = sales_state, x = 'State',y= 'Amount') #this will plot the bar chart with the top 10 states of the amount list
```

*From above graphs we can see that most of the orders & total sales/amount are from Uttar Pradesh, Maharashtra and Karnataka respectively*

```python
### Marital Status

ax = sns.countplot(data = df, x = 'Marital_Status') #this will plot the count of the marital status

sns.set(rc={'figure.figsize':(7,5)}) #this will set the figure size of the plot
for bars in ax.containers: #this will add the count labels
    ax.bar_label(bars) #this will add the count labels

sales_state = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False) #this will group the data by marital

sns.set(rc={'figure.figsize':(6,5)}) #this will set the figure size of the plot
sns.barplot(data = sales_state, x = 'Marital_Status',y= 'Amount', hue='Gender') #this will plot the bar chart with the marital status 
```

*From above graphs we can see that most of the buyers are married (women) and they have high purchasing power*

```python
### Occupation

sns.set(rc={'figure.figsize':(20,5)}) #this will set the figure size of the plot
ax = sns.countplot(data = df, x = 'Occupation') #this will plot the count of the occupation 
 
for bars in ax.containers:  #this will add the count labels 
    ax.bar_label(bars) 

sales_state = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False) #this will group the data by occupation and sum the amount

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Occupation',y= 'Amount')
```

*From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector*

```python
### Product Category

sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Product_Category')

for bars in ax.containers:
    ax.bar_label(bars)

sales_state = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10) #this will group the data by product category and sum the amount

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_Category',y= 'Amount')
```

*From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category*

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10) #this will group the data by product id and sum the orders

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_ID',y= 'Orders')
```

```python
# top 10 most sold products 

fig1, ax1 = plt.subplots(figsize=(12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar') #this will plot the top 10 most sold products 
```
