### Geographic analysis

### 1. Plot the locations of restaurants on map using longitude and latitude coordinates.

### 2. Identify  any patterns or clusters of restaurants in specific area.

```python
# Importing import libraries
import folium
from folium.plugins import HeatMap
import pandas as pd
```

```python
# Reading the data into the file
Data = pd.read_csv("Dataset .csv")
```

```python
# Types of data 
Data.info()
```

### Plot the locations of restaurants on map using longitude and latitude coordinates.

```python
# Create a map centered at an average location
map_center = [Data["Latitude"].mean(), Data["Longitude"].mean()]
restaurant_map = folium.Map(location=map_center, zoom_start=10)

# Add restaurant locations to the map
for index, row in Data.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=row["Restaurant Name"],
        icon=folium.Icon(color="red", icon="cutlery"),
    ).add_to(restaurant_map)

# Display the map
restaurant_map
```

### Plot the locations of restaurants on heatmap using longitude and latitude coordinates and  Identify any patterns or clusters of restaurants in specific area.


```python
# Create a map centered at an average location
map_center = [Data["Latitude"].mean(), Data["Longitude"].mean()]
restaurant_map = folium.Map(location=map_center, zoom_start=10)
heat_data = Data[["Latitude", "Longitude"]].values.tolist()  # Convert to list of coordinates

HeatMap(heat_data, radius=15, blur=10, max_zoom=13).add_to(restaurant_map)


# Display the map
restaurant_map
```
