### Restaurants Reviews

### 1. Analyze the text reviews to identify the most common positive and negative keywords.

### 2. Calculate the average length of reviews and explore if there is relationship between review length and rating.

```python
# importind required libraries
import pandas as pd
```

```python
# load data into the file 
Data = pd.read_csv("Dataset .csv")
```

```python
# type of data
Data.info()
```

### Analyze the text reviews to identify the most common positive and negative keywords.

```python
# Occurrence of a particular reviews.
Data["Rating text"].value_counts()
```

```python
# Average votes on a particular review
Data.columns = Data.columns.str.strip()
Text_Reviews = Data.groupby("Rating text")["Votes"].mean().reset_index()
Text_Reviews = Text_Reviews.sort_values(by=["Votes"], ascending=False)
Text_Reviews
```

### Calculate the average length of reviews and explore if there is relationship between review length and rating.

```python
# Calculate the length of each review
Data.columns = Data.columns.str.strip()
Data["Rating text"] = Data["Rating text"].astype(str).apply(len)

# Calculate the average length
average_length = Data["Rating text"].mean()
average_length.round(2)
```

###  Relationship between review length and rating.

```python
# Removing the white space
Data.columns = Data.columns.str.strip()

# Grouping "Rating text" categories (e.g., "Excellent", "Good", "Average")
# and calculating the average "Aggregate rating" for each category.
Relationship = Data.groupby("Rating text")["Aggregate rating"].mean().reset_index()
Relationship
```

```python
# the correlation between Rating text and aggregate rating.
Data[["Rating text", "Aggregate rating"]].corr()
```
